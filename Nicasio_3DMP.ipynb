###  Draw the following in Python and compute for the volume of each 3D image (all edge color must be red).

#### a. Cylinder whose radius = 13.3, height = 17 and number of points/resolution = 450 (yellow)

```python
import numpy as np
import matplotlib.pyplot as plt

def draw_cylinder(ax, radius, height, resolution, color, edge_color):
    z = np.linspace(0, height, 50)
    theta = np.linspace(0, 2 * np.pi, resolution)
    theta_grid, z_grid = np.meshgrid(theta, z)
    x = radius * np.cos(theta_grid)
    y = radius * np.sin(theta_grid)
    ax.plot_surface(x, y, z_grid, color=color, edgecolor=edge_color, alpha=0.8)

    return np.pi * radius**2 * height 


fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
vol_cylinder = draw_cylinder(ax, 13.3, 17, 450, 'yellow', 'red')
ax.set_title("Cylinder")
plt.show()

print("Volume of Cylinder:", vol_cylinder)
```

### b. Cone whose radius = 13.3, height = 17 and number of points/resolution = 450 (cyan)

```python
import numpy as np
import matplotlib.pyplot as plt

cone_radius = 13.3
cone_height = 17
resolution = 450

theta = np.linspace(0, 2*np.pi, resolution)
z = np.linspace(0, cone_height, resolution)
theta, Z = np.meshgrid(theta, z)

X = cone_radius * (1 - Z/cone_height) * np.cos(theta)
Y = cone_radius * (1 - Z/cone_height) * np.sin(theta)

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')


ax.plot_surface(X, Y, Z, rstride=5, cstride=5, color='cyan', edgecolor='red')


ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
plt.show()

cone_volume = (1/3) * np.pi * cone_radius**2 * cone_height
print(f"Volume of the cone: {cone_volume}")
```

### c. Sphere whose radius = 13.3 and number of points/resolution = 450 (magenta)

```python
import numpy as np
import matplotlib.pyplot as plt

def draw_sphere(ax, radius, resolution, color, edge_color):
    phi = np.linspace(0, np.pi, resolution)
    theta = np.linspace(0, 2 * np.pi, resolution)
    phi_grid, theta_grid = np.meshgrid(phi, theta)
    x = radius * np.sin(phi_grid) * np.cos(theta_grid)
    y = radius * np.sin(phi_grid) * np.sin(theta_grid)
    z = radius * np.cos(phi_grid)
    ax.plot_surface(x, y, z, color=color, edgecolor=edge_color, alpha=0.8)

    return (4/3) * np.pi * radius**3  

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
vol_sphere = draw_sphere(ax, 13.3, 450, 'magenta', 'red')
ax.set_title("Sphere")
plt.show()

print("Volume of Sphere:", vol_sphere)
```

### d. Cube whose initial coordinate located at (4, 8, 9) and the edge is 12. (blue)

```python
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

initial_coord = (4, 8, 9)
edge_length = 12

vertices = np.array([
    [initial_coord[0], initial_coord[1], initial_coord[2]],  # Corner 1
    [initial_coord[0] + edge_length, initial_coord[1], initial_coord[2]],  # Corner 2
    [initial_coord[0] + edge_length, initial_coord[1] + edge_length, initial_coord[2]],  # Corner 3
    [initial_coord[0], initial_coord[1] + edge_length, initial_coord[2]],  # Corner 4
    [initial_coord[0], initial_coord[1], initial_coord[2] + edge_length],  # Corner 5
    [initial_coord[0] + edge_length, initial_coord[1], initial_coord[2] + edge_length],  # Corner 6
    [initial_coord[0] + edge_length, initial_coord[1] + edge_length, initial_coord[2] + edge_length],  # Corner 7
    [initial_coord[0], initial_coord[1] + edge_length, initial_coord[2] + edge_length]   # Corner 8
])

faces = [
    [vertices[0], vertices[1], vertices[2], vertices[3]],  # Bottom
    [vertices[4], vertices[5], vertices[6], vertices[7]],  # Top
    [vertices[0], vertices[1], vertices[5], vertices[4]],  # Front
    [vertices[2], vertices[3], vertices[7], vertices[6]],  # Back
    [vertices[1], vertices[2], vertices[6], vertices[5]],  # Right
    [vertices[4], vertices[7], vertices[3], vertices[0]]   # Left
]

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.add_collection3d(Poly3DCollection(faces, facecolors='blue', linewidths=1, edgecolors='red', alpha=.30))

ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')


ax.set_xlim([initial_coord[0], initial_coord[0] + edge_length])
ax.set_ylim([initial_coord[1], initial_coord[1] + edge_length])
ax.set_zlim([initial_coord[2], initial_coord[2] + edge_length])


plt.show()


cube_volume = edge_length ** 3
print(f"Volume of the cube: {cube_volume}")

```

### e. Pyramid whose base is rectangle (L = 21 and W = 16) and height is 23 units. (green)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection


def draw_pyramid(ax, base_dims, height, color, edge_color):
    L, W = base_dims
 
    vertices = [
        [0, 0, 0],          
        [L, 0, 0],           
        [L, W, 0],           
        [0, W, 0],           
        [L / 2, W / 2, height],  
    ]

    faces = [
        [vertices[0], vertices[1], vertices[4]],  
        [vertices[1], vertices[2], vertices[4]],  
        [vertices[2], vertices[3], vertices[4]],  
        [vertices[3], vertices[0], vertices[4]],  
        [vertices[0], vertices[1], vertices[2], vertices[3]], 
    ]
   
    poly3d = Poly3DCollection(faces, facecolors=color, edgecolors=edge_color, alpha=0.8)
    ax.add_collection3d(poly3d)
   
    for face in faces:
        ax.plot([v[0] for v in face + [face[0]]],
                [v[1] for v in face + [face[0]]],
                [v[2] for v in face + [face[0]]], color=edge_color)

    
    return (1/3) * L * W * height


fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
vol_pyramid = draw_pyramid(ax, (21, 16), 23, 'green', 'red')
ax.set_title("Pyramid")
ax.set_xlim([0, 25])
ax.set_ylim([0, 25])
ax.set_zlim([0, 30])
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
plt.show()

print("Volume of Pyramid:", vol_pyramid)
```

### f. Prism whose base is rectangle (L = 21 and W = 16) and height is 23 units. (orange)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

def draw_prism(ax, base_dims, height, color, edge_color):
    L, W = base_dims
   
    vertices = [
        [0, 0, 0],          # Bottom-front-left corner
        [L, 0, 0],          # Bottom-front-right corner
        [L, W, 0],          # Bottom-back-right corner
        [0, W, 0],          # Bottom-back-left corner
        [0, 0, height],     # Top-front-left corner
        [L, 0, height],     # Top-front-right corner
        [L, W, height],     # Top-back-right corner
        [0, W, height],     # Top-back-left corner
    ]
   
    faces = [
        [vertices[0], vertices[1], vertices[5], vertices[4]],  # Front face
        [vertices[1], vertices[2], vertices[6], vertices[5]],  # Right face
        [vertices[2], vertices[3], vertices[7], vertices[6]],  # Back face
        [vertices[3], vertices[0], vertices[4], vertices[7]],  # Left face
        [vertices[4], vertices[5], vertices[6], vertices[7]],  # Top face
        [vertices[0], vertices[1], vertices[2], vertices[3]],  # Bottom face
    ]
    
    poly3d = Poly3DCollection(faces, facecolors=color, edgecolors=edge_color, alpha=0.8)
    ax.add_collection3d(poly3d)
    
    for face in faces:
        ax.plot([v[0] for v in face + [face[0]]],
                [v[1] for v in face + [face[0]]],
                [v[2] for v in face + [face[0]]], color=edge_color)

   
    return L * W * height


fig = plt.figure(figsize=(8, 8))
ax = fig.add_subplot(111, projection='3d')
vol_prism = draw_prism(ax, (21, 16), 23, 'orange', 'red')
ax.set_title("Prism")
ax.set_xlim([0, 30])
ax.set_ylim([0, 25])
ax.set_zlim([0, 30])
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
plt.show()

print("Volume of Prism:", vol_prism)
```
