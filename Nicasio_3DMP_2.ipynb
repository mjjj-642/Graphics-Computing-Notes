### 2.) Find the distance of the following: a. X to Y and (b) X to Z where X(2, 3, 6), Y(7, 8, 9), and Z (2, 9, 4).

```python
import numpy as np

X = np.array([2, 3, 6])
Y = np.array([7, 8, 9])
Z = np.array([2, 9, 4])

# Compute distances
distance_XY = np.linalg.norm(X - Y)
distance_XZ = np.linalg.norm(X - Z)

print("Distance from X to Y:", distance_XY)
print("Distance from X to Z:", distance_XZ)
```
