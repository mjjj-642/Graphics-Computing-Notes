### 1.) Create a 3D line plot that shows the path of a sine wave in 3D space.

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

t = np.linspace(0, 20, 500)  
x = np.sin(t)  
y = np.cos(t) 
z = t          

# Create the plot
fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot(x, y, z, label="3D Sine Wave Path", color='blue', linewidth=2)
ax.set_title("3D Line Plot of Sine Wave Path")
ax.set_xlabel("X-axis (sin(t))")
ax.set_ylabel("Y-axis (cos(t))")
ax.set_zlabel("Z-axis (t)")

ax.legend()
plt.show()
```
