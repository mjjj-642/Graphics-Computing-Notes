### 1.) (a) spheres whose radius is 25 and 300 resolution

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the radius and resolution
radius = 25
resolution = 300

# Create a grid of points
phi = np.linspace(0, np.pi, resolution)
theta = np.linspace(0, 2 * np.pi, resolution)
phi, theta = np.meshgrid(phi, theta)

# Convert spherical coordinates to Cartesian coordinates
x = radius * np.sin(phi) * np.cos(theta)
y = radius * np.sin(phi) * np.sin(theta)
z = radius * np.cos(phi)

# Plot the sphere
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(x, y, z, color='b', alpha=0.6, edgecolor='r')

# Set the aspect ratio to be equal
ax.set_box_aspect([1, 1, 1])

# Set labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')

# Set title
ax.set_title('Sphere with Radius 25 and Resolution 300')

plt.show()
```

### 2.) (c) Scale the sphere using the given parameters c.1) (Sx, Sy, Sz) = (0.25, 4, 1) , c.2) (Sx, Sy, Sz) = (2, 5, -1)

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the radius and resolution
radius = 25
resolution = 300

# Create a grid of points
phi = np.linspace(0, np.pi, resolution)
theta = np.linspace(0, 2 * np.pi, resolution)
phi, theta = np.meshgrid(phi, theta)

# Convert spherical coordinates to Cartesian coordinates
x = radius * np.sin(phi) * np.cos(theta)
y = radius * np.sin(phi) * np.sin(theta)
z = radius * np.cos(phi)

# Function to plot scaled spheres in a single figure
def plot_scaled_spheres(x, y, z, scaling_params):
    fig = plt.figure(figsize=(12, 6))

    for i, (Sx, Sy, Sz, title) in enumerate(scaling_params, 1):
        # Apply scaling
        x_scaled = Sx * x
        y_scaled = Sy * y
        z_scaled = Sz * z

        # Create subplot
        ax = fig.add_subplot(1, len(scaling_params), i, projection='3d')
        ax.plot_surface(x_scaled, y_scaled, z_scaled, color='b', alpha=0.6, edgecolor='r')

        # Set the aspect ratio to be equal
        ax.set_box_aspect([1, 1, 1])

        # Set labels
        ax.set_xlabel('X')
        ax.set_ylabel('Y')
        ax.set_zlabel('Z')

        # Set title
        ax.set_title(title)

    plt.tight_layout()
    plt.show()

# Scaling parameters
scaling_params = [
    (0.25, 4, 1, "Scaled Sphere: Sx=0.25, Sy=4, Sz=1"),
    (2, 5, -1, "Scaled Sphere: Sx=2, Sy=5, Sz=-1")
]

# Plot the scaled spheres
plot_scaled_spheres(x, y, z, scaling_params)
```

#### (Sx, Sy, Sz) = (0.25, 4, 1) Rotated in Respect to X, Y, Z axis 

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the radius and resolution
radius = 25
resolution = 300

# Create a grid of points
phi = np.linspace(0, np.pi, resolution)
theta = np.linspace(0, 2 * np.pi, resolution)
phi, theta = np.meshgrid(phi, theta)

# Convert spherical coordinates to Cartesian coordinates
x = radius * np.sin(phi) * np.cos(theta)
y = radius * np.sin(phi) * np.sin(theta)
z = radius * np.cos(phi)

# Function to apply rotation around an axis
def rotate_coordinates(x, y, z, angle, axis):
    angle_rad = np.radians(angle)
    if axis == 'x':
        y_rot = y * np.cos(angle_rad) - z * np.sin(angle_rad)
        z_rot = y * np.sin(angle_rad) + z * np.cos(angle_rad)
        return x, y_rot, z_rot
    elif axis == 'y':
        x_rot = x * np.cos(angle_rad) + z * np.sin(angle_rad)
        z_rot = -x * np.sin(angle_rad) + z * np.cos(angle_rad)
        return x_rot, y, z_rot
    elif axis == 'z':
        x_rot = x * np.cos(angle_rad) - y * np.sin(angle_rad)
        y_rot = x * np.sin(angle_rad) + y * np.cos(angle_rad)
        return x_rot, y_rot, z
    else:
        raise ValueError("Invalid axis. Choose from 'x', 'y', or 'z'.")

# Function to plot original and rotated scaled sphere
def plot_rotated_scaled_sphere(x, y, z, Sx, Sy, Sz, angle):
    # Apply scaling
    x_scaled = Sx * x
    y_scaled = Sy * y
    z_scaled = Sz * z

    # Rotate scaled sphere
    x_rot_x, y_rot_x, z_rot_x = rotate_coordinates(x_scaled, y_scaled, z_scaled, angle, 'x')
    x_rot_y, y_rot_y, z_rot_y = rotate_coordinates(x_scaled, y_scaled, z_scaled, angle, 'y')
    x_rot_z, y_rot_z, z_rot_z = rotate_coordinates(x_scaled, y_scaled, z_scaled, angle, 'z')

    # Plot the spheres
    fig = plt.figure(figsize=(16, 8))

    # Original scaled sphere
    ax1 = fig.add_subplot(1, 4, 1, projection='3d')
    ax1.plot_surface(x_scaled, y_scaled, z_scaled, color='b', alpha=0.6, edgecolor='r')
    ax1.set_title("Original Scaled Sphere (Sx, Sy, Sz) = (0.25, 4, 1)")
    ax1.set_box_aspect([1, 1, 1])

    # Rotated about X-axis
    ax2 = fig.add_subplot(1, 4, 2, projection='3d')
    ax2.plot_surface(x_rot_x, y_rot_x, z_rot_x, color='g', alpha=0.6, edgecolor='r')
    ax2.set_title("Rotated 75° about X-axis")
    ax2.set_box_aspect([1, 1, 1])

    # Rotated about Y-axis
    ax3 = fig.add_subplot(1, 4, 3, projection='3d')
    ax3.plot_surface(x_rot_y, y_rot_y, z_rot_y, color='c', alpha=0.6, edgecolor='r')
    ax3.set_title("Rotated 75° about Y-axis")
    ax3.set_box_aspect([1, 1, 1])

    # Rotated about Z-axis
    ax4 = fig.add_subplot(1, 4, 4, projection='3d')
    ax4.plot_surface(x_rot_z, y_rot_z, z_rot_z, color='m', alpha=0.6, edgecolor='r')
    ax4.set_title("Rotated 75° about Z-axis")
    ax4.set_box_aspect([1, 1, 1])

    plt.tight_layout()
    plt.show()

# Scaling and rotation parameters
Sx, Sy, Sz = 0.25, 4, 1
angle = 75

# Plot the rotated scaled sphere
plot_rotated_scaled_sphere(x, y, z, Sx, Sy, Sz, angle)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Define the radius and resolution
radius = 25
resolution = 300

# Create a grid of points
phi = np.linspace(0, np.pi, resolution)
theta = np.linspace(0, 2 * np.pi, resolution)
phi, theta = np.meshgrid(phi, theta)

# Convert spherical coordinates to Cartesian coordinates
x = radius * np.sin(phi) * np.cos(theta)
y = radius * np.sin(phi) * np.sin(theta)
z = radius * np.cos(phi)

# Function to apply rotation around an axis
def rotate_coordinates(x, y, z, angle, axis):
    angle_rad = np.radians(angle)
    if axis == 'x':
        y_rot = y * np.cos(angle_rad) - z * np.sin(angle_rad)
        z_rot = y * np.sin(angle_rad) + z * np.cos(angle_rad)
        return x, y_rot, z_rot
    elif axis == 'y':
        x_rot = x * np.cos(angle_rad) + z * np.sin(angle_rad)
        z_rot = -x * np.sin(angle_rad) + z * np.cos(angle_rad)
        return x_rot, y, z_rot
    elif axis == 'z':
        x_rot = x * np.cos(angle_rad) - y * np.sin(angle_rad)
        y_rot = x * np.sin(angle_rad) + y * np.cos(angle_rad)
        return x_rot, y_rot, z
    else:
        raise ValueError("Invalid axis. Choose from 'x', 'y', or 'z'.")

# Function to plot original and rotated scaled sphere
def plot_rotated_scaled_sphere(x, y, z, Sx, Sy, Sz, angle):
    # Apply scaling
    x_scaled = Sx * x
    y_scaled = Sy * y
    z_scaled = Sz * z

    # Rotate scaled sphere
    x_rot_x, y_rot_x, z_rot_x = rotate_coordinates(x_scaled, y_scaled, z_scaled, angle, 'x')
    x_rot_y, y_rot_y, z_rot_y = rotate_coordinates(x_scaled, y_scaled, z_scaled, angle, 'y')
    x_rot_z, y_rot_z, z_rot_z = rotate_coordinates(x_scaled, y_scaled, z_scaled, angle, 'z')

    # Plot the spheres
    fig = plt.figure(figsize=(16, 8))

    # Original scaled sphere
    ax1 = fig.add_subplot(1, 4, 1, projection='3d')
    ax1.plot_surface(x_scaled, y_scaled, z_scaled, color='b', alpha=0.6, edgecolor='r')
    ax1.set_title("Original Scaled Sphere")
    ax1.set_box_aspect([1, 1, 1])

    # Rotated about X-axis
    ax2 = fig.add_subplot(1, 4, 2, projection='3d')
    ax2.plot_surface(x_rot_x, y_rot_x, z_rot_x, color='g', alpha=0.6, edgecolor='r')
    ax2.set_title("Rotated 75° about X-axis")
    ax2.set_box_aspect([1, 1, 1])

    # Rotated about Y-axis
    ax3 = fig.add_subplot(1, 4, 3, projection='3d')
    ax3.plot_surface(x_rot_y, y_rot_y, z_rot_y, color='c', alpha=0.6, edgecolor='r')
    ax3.set_title("Rotated 75° about Y-axis")
    ax3.set_box_aspect([1, 1, 1])

    # Rotated about Z-axis
    ax4 = fig.add_subplot(1, 4, 4, projection='3d')
    ax4.plot_surface(x_rot_z, y_rot_z, z_rot_z, color='m', alpha=0.6, edgecolor='r')
    ax4.set_title("Rotated 75° about Z-axis")
    ax4.set_box_aspect([1, 1, 1])

    plt.tight_layout()
    plt.show()

# Scaling and rotation parameters
Sx, Sy, Sz = 2, 5, -1
angle = 75

# Plot the rotated scaled sphere
plot_rotated_scaled_sphere(x, y, z, Sx, Sy, Sz, angle)
```

### Observed Comparison of (Sx, Sy, Sz) = (0.25, 4, 1) and (Sx, Sy, Sz) = (2, 5, -1)

    - In the in respect of Y-Axis (0.25, 4, 1) is a lot more slimer and it slanted to the right, while (2,5,-1) is a little bit wider and slanted to the left
    - In the in respect of Z-Axis (0.25, 4, 1) is way more slimer, while (2,5,-1) is also a little bit wider compared to the other. Both objects are a little slanted to left
